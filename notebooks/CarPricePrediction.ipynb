# Feature Engineering

```python
df["km_per_year"] = df["km_driven"] / (df["vehicle_age"] + 1)

df["power_per_litre"] = df["max_power"] / df["engine"]

df["engine_per_seat"] = df["engine"] / df["seats"]

df["age_km_interaction"] = df["vehicle_age"] * df["km_driven"]

df["high_mileage_vehicle"] = (df["km_driven"] > 100000).astype(int)
```
